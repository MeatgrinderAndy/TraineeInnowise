**1.** Дан случайный массив, поменять знак у элементов, значения которых между 3 и 8

```python
array = np.random.randint(1, 10, size=(10))
print(array)

array[(array>3) & (array<8)] *= -1
print(array)
```

**2.** Заменить максимальный элемент случайного массива на 0

```python
array = np.random.randint(1, 10, size=(10))
print(array)

array[array == np.max(array)] = 0
print(array)
```

**3.** Построить прямое произведение массивов (все комбинации с каждым элементом). На вход подается двумерный массив

```python
array = np.random.randint(1, 10, size=(3, 3))
print(array)

kron_array = np.kron(array, array)
print(kron_array)
```

**4.** Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
arrayA = np.random.randint(1, 10, size=(8, 3))
arrayB = np.random.randint(1, 10, size=(2, 2))
print(arrayA, '\n\n', arrayB)

rowIndexes = []

arrayB_elements = arrayB.flatten()
print(arrayB_elements)

for index, row in enumerate(arrayA):
  if np.all(np.isin(arrayB_elements, row)):
    rowIndexes.append(index)

arrayA_rows = arrayA[rowIndexes]
print(arrayA_rows)
```

**5.** Дана 10x3 матрица, найти строки из неравных значений (например строка [2,2,3] остается, строка [3,3,3] удаляется)

```python
array = np.random.randint(1, 5, size=(10, 3))
print(array)

rowIndexes = []

for index, row in enumerate(array):
  if np.all(np.diff(row)==0):
    rowIndexes.append(index)

print(np.delete(array, rowIndexes, axis=0))
```

**6.** Дан двумерный массив. Удалить те строки, которые повторяются

```python
array = np.random.randint(1, 5, size=(10, 3))
print(array)

unique_array = np.unique(array, axis=0)
print(unique_array)
```

* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
X = np.array([[1, 0, 1],
              [2, 0, 2],
              [3, 0, 3],
              [4, 4, 4]])

diagonalX = np.diag(X)
result = np.prod(diagonalX[diagonalX != 0])
print(result)

#############

X_list = [[1, 0, 1],
          [2, 0, 2],
          [3, 0, 3],
          [4, 4, 4]]

diagonal_length = min(len(X_list), len(X_list[0])) if len(X_list) != 0 else 0
non_zero_elements = False
result = 1


for i in range(diagonal_length):
  if X_list[i][i] != 0:
    result *= X_list[i][i]
    non_zero_elements = True
if non_zero_elements == False:
  result = 0

print(result)
```

* __Задача 3__: Найти максимальный элемент в векторе x среди элементов, перед которыми стоит ноль.

```python
x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])

zero_indexes = np.where(x == 0)[0]
past_zeroes_indexes = zero_indexes + 1
x_past_zeroes = x[past_zeroes_indexes[past_zeroes_indexes < len(x)]]
max_value_past_zero = np.max(x_past_zeroes)
print(max_value_past_zero)

#########

x_list = [6, 2, 0, 3, 0, 0, 5, 7, 0]

current_max = 0

for i in range(len(x_list)):
  if i>0 and x_list[i-1] == 0 and x_list[i] > current_max:
    current_max = x_list[i]

print(current_max)
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
x = np.array([2, 2, 2, 3, 3, 3, 5])
edges = np.where(x[1:] != x[:-1])[0]
edges = np.concatenate((edges, [len(x) - 1]))
values = x[edges]
counts = np.diff(np.concatenate(([-1], edges)))
result = [values, counts]
print(result)

#######

x_list = [2, 2, 2, 3, 3, 3, 5]
values = []
counts = []
for i in range(len(x_list)):
  if i==len(x_list)-1 or (i<len(x_list)-1 and x_list[i] != x_list[i+1]):
    values.append(x_list[i])
    counts.append(i+1 if len(counts)==0 else (i + 1 - sum(counts)))

result = [values, counts]
print(result)
```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
from scipy.spatial.distance import cdist
import time

X = np.random.rand(1000, 200)
Y = np.random.rand(2000, 200)

def euclidean_distance(X, Y):
    x2 = np.sum(X**2, axis=1)[:, np.newaxis]
    y2 = np.sum(Y**2, axis=1)[np.newaxis, :]
    xy = np.dot(X, Y.T)
    sq_dists = np.maximum(x2 - 2 * xy + y2, 0)
    return np.sqrt(sq_dists)

start = time.time()
distance_numpy = euclidean_distance(X, Y)
result_time_numpy = time.time() - start
print(f'Numpy time: {result_time_numpy}')

start = time.time()
dist_scipy = cdist(X, Y, metric='euclidean')
result_time_scipy = time.time() - start
print(f"SciPy time: {result_time_scipy}")
```
