```python
# загружаем датасет
data = pd.read_csv("./data/adult.data.csv")
data.head()
```

**2. Каков средний возраст мужчин (признак *age*) по всему датасету?**

**3. Какова доля граждан Соединенных Штатов (признак *native-country*)?**

**4-5. Рассчитайте среднее значение и среднеквадратичное отклонение возраста тех, кто получает более 50K в год (признак *salary*) и тех, кто получает менее 50K в год**

```python
less_than_50k_ages = data[data["salary"] == "<=50K"]["age"]
more_than_50k_ages = data[data["salary"] == ">50K"]["age"]

print(
f'''Mean Age:
    Salary more than 50k: {more_than_50k_ages.mean()}
    Salary less than 50k: {less_than_50k_ages.mean()}
Age Stds:
    Salary more than 50k: {more_than_50k_ages.std()}
    Salary less than 50k: {less_than_50k_ages.std()}''')
```

**6. Правда ли, что люди, которые получают больше 50k, имеют минимум высшее образование? (признак *education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters* или *Doctorate*)**

```python
more_than_50k_education = data[data["salary"] == ">50K"]["education"]
more_than_50k_education = more_than_50k_education.value_counts(normalize=True)
percentage_of_education = more_than_50k_education[["Bachelors", "Prof-school", "Assoc-acdm", "Assoc-voc", "Masters", "Doctorate"]].sum() * 100
print(f'Only {round(percentage_of_education, 2)}% of rich people have higher education')
```

**7. Выведите статистику возраста для каждой расы (признак *race*) и каждого пола. Используйте *groupby* и *describe*. Найдите таким образом максимальный возраст мужчин расы *Asian-Pac-Islander*.**

```python
age_stats = data.groupby(["race", "sex"])["age"].describe()
max_age = age_stats.loc[('Asian-Pac-Islander', 'Male'), 'max']
max_age
```

**8. Среди кого больше доля зарабатывающих много (>50K): среди женатых или холостых мужчин (признак *marital-status*)? Женатыми считаем тех, у кого *marital-status* начинается с *Married* (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), остальных считаем холостыми.**

```python
rich_men_marital_statuses = data[(data['sex'] == 'Male') & (data['salary'] == '>50K')]['marital-status']
married_rich = rich_men_marital_statuses.value_counts(normalize=True)[['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse']].sum()
married_rich_percentage = round(married_rich* 100, 2)
print(f'''Married percentage: {married_rich_percentage}%
Not married percentage: {round(100 - married_rich_percentage, 2)}%''')
```

**9. Какое максимальное число часов человек работает в неделю (признак *hours-per-week*)? Сколько людей работают такое количество часов и каков среди них процент зарабатывающих много?**

```python
max_hours = data['hours-per-week'].max()
hardworking_people = data[data['hours-per-week'] == max_hours]
hardworking_rich_people = hardworking_people[hardworking_people['salary'] == '>50K']
print(f'''Number of people who work {max_hours} a week: {len(hardworking_people)}
Number of these people who are rich: {len(hardworking_rich_people)}
Percentage of rich {round(len(hardworking_rich_people)/len(hardworking_people) * 100, 2)}%''')
```

**10. Посчитайте среднее время работы (*hours-per-week*) зарабатывающих мало и много (*salary*) для каждой страны (*native-country*).**

**11.Сгруппируйте людей по возрастным группам *young*, *adult*, *retiree*, где:**
* *young* соответствует 16-35 лет
* *adult* - 35-70 лет
* *retiree* - 70-100 лет

**Проставьте название соответсвтуещей группы для каждого человека в новой колонке AgeGroup**

```python
data['AgeGroup'] = pd.cut(data['age'], bins=[16, 35, 70, 100], labels=['young', 'adult', 'retiree'])
data
```

**12-13. Определите количество зарабатывающих >50K в каждой из возрастных групп (колонка AgeGroup), а также выведите название возрастной группы, в которой чаще зарабатывают больше 50К (>50K)**

```python
rich_groups = data[data['salary'] == '>50K']['AgeGroup'].value_counts()
print(f'Most common to be rich is "{rich_groups.idxmax()}" group')
```

**14. Сгруппируйте людей по типу занятости (колонка occupation) и определите количество людей в каждой группе. После чего напишите функциюю фильтрации filter_func, которая будет возвращать только те группы, в которых средний возраст (колонка age) не больше 40 и в которых все работники отрабатывают более 5 часов в неделю (колонка hours-per-week)**

```python
def group_filter_func(rows):
  return rows['age'].mean() <= 40 and (rows['hours-per-week'] > 5).all()

result = data.groupby(['occupation']).filter(group_filter_func)
print(result['occupation'].unique())
```
